```python
# Data komoditas
nama_komoditas = "Beras"
harga_per_kg = 15000
stok_kg = 100

def hitung_nilai_stok(harga, stok):
    return harga * stok

def tambah_stok(stok, jumlah):
    return stok + jumlah

print("Nama:", nama_komoditas)
print("Nilai stok:", hitung_nilai_stok(
    harga_per_kg, stok_kg
))

stok_kg = tambah_stok(stok_kg, 20)
print("Stok setelah ditambah:", stok_kg)
```

```python
class KomoditasPertanian:
    """Menyimpan data dan mengelola komoditas pertanian."""

    def __init__(self, nama, harga_per_kg, stok_kg):
        """Menginisialisasi data komoditas."""
        self.nama = nama
        self.harga_per_kg = harga_per_kg
        self.stok_kg = stok_kg

    def hitung_nilai_stok(self):
        """Menghitung nilai seluruh stok komoditas."""
        return self.harga_per_kg * self.stok_kg

    def tambah_stok(self, jumlah):
        """Menambahkan jumlah stok komoditas."""
        self.stok_kg += jumlah


beras = KomoditasPertanian("Beras", 15000, 100)

print("Nama:", beras.nama)
print("Nilai stok:", beras.hitung_nilai_stok())

beras.tambah_stok(20)
print("Stok setelah ditambah:", beras.stok_kg)
```

```python

class KomoditasPertanian:
    """Merepresentasikan komoditas pertanian dan persediaannya."""

    jumlah_objek = 0
    diskon_persen = 0.0

    def __init__(self, nama: str, harga_per_kg: float,
                 stok_kg: float):
        """Membuat objek komoditas dengan validasi data."""
        if not nama.strip():
            raise ValueError("Nama komoditas tidak boleh kosong.")
        if harga_per_kg < 0:
            raise ValueError("Harga tidak boleh negatif.")
        if stok_kg < 0:
            raise ValueError("Stok tidak boleh negatif.")

        self.nama = nama
        self.harga_per_kg = harga_per_kg
        self.stok_kg = stok_kg
        KomoditasPertanian.jumlah_objek += 1

    def hitung_nilai_stok(self) -> float:
        """Mengembalikan nilai seluruh stok sebelum diskon."""
        return self.harga_per_kg * self.stok_kg

    def hitung_harga_diskon(self) -> float:
        """Menghitung harga per kg setelah diskon kelas."""
        return self.harga_per_kg * (
            1 - KomoditasPertanian.diskon_persen / 100
        )

    def tambah_stok(self, jumlah: float) -> None:
        """Menambahkan stok dengan jumlah yang valid."""
        if jumlah <= 0:
            raise ValueError("Jumlah penambahan harus positif.")
        self.stok_kg += jumlah

    def jual(self, jumlah: float) -> None:
        """Mengurangi stok jika persediaan mencukupi."""
        if jumlah <= 0:
            raise ValueError("Jumlah penjualan harus positif.")
        if jumlah > self.stok_kg:
            raise ValueError("Stok tidak mencukupi.")
        self.stok_kg -= jumlah

    def __str__(self) -> str:
        """Menghasilkan ringkasan objek dalam bentuk teks."""
        return (
            f"{self.nama}: stok {self.stok_kg} kg, "
            f"harga Rp{self.harga_per_kg:,.0f}/kg"
        )


class Gudang:
    """Mengelola penyimpanan komoditas dalam batas kapasitas."""

    def __init__(self, nama: str, kapasitas_kg: float):
        """Membuat gudang dengan kapasitas yang valid."""
        if not nama.strip():
            raise ValueError("Nama gudang tidak boleh kosong.")
        if kapasitas_kg <= 0:
            raise ValueError("Kapasitas harus lebih dari nol.")

        self.nama = nama
        self.kapasitas_kg = kapasitas_kg
        self.daftar_komoditas = []

    def hitung_total_stok(self) -> float:
        """Menghitung total stok seluruh komoditas di gudang."""
        return sum(
            komoditas.stok_kg
            for komoditas in self.daftar_komoditas
        )

    def simpan_komoditas(self, komoditas: KomoditasPertanian):
        """Menyimpan komoditas jika kapasitas mencukupi."""
        if komoditas in self.daftar_komoditas:
            raise ValueError("Komoditas sudah terdaftar.")

        total_baru = self.hitung_total_stok() + komoditas.stok_kg
        if total_baru > self.kapasitas_kg:
            raise ValueError("Kapasitas gudang tidak mencukupi.")

        self.daftar_komoditas.append(komoditas)
        return (
            f"{komoditas.nama} berhasil disimpan di {self.nama}."
        )


# LO 3: Membuat empat objek dengan data berbeda
komoditas_list = [
    KomoditasPertanian("Beras", 15000, 100),
    KomoditasPertanian("Jagung", 8000, 150),
    KomoditasPertanian("Kedelai", 12000, 80),
    KomoditasPertanian("Kentang", 10000, 60)
]

# Mengakses atribut menggunakan notasi titik
print("=== DATA KOMODITAS ===")
for komoditas in komoditas_list:
    print("Nama:", komoditas.nama)
    print("Harga/kg:", komoditas.harga_per_kg)
    print("Stok:", komoditas.stok_kg)
    print("Nilai stok:", komoditas.hitung_nilai_stok())
    print(komoditas)
    print()

# LO 3: Interaksi antarobjek
gudang = Gudang("Gudang Utama", 500)

print("=== INTERAKSI OBJEK ===")
print(gudang.simpan_komoditas(komoditas_list[0]))
print(gudang.simpan_komoditas(komoditas_list[1]))
print("Total stok gudang:", gudang.hitung_total_stok(), "kg")

# Menambahkan dan menjual stok
komoditas_list[0].tambah_stok(20)
print("Stok beras setelah penambahan:",
      komoditas_list[0].stok_kg, "kg")

komoditas_list[0].jual(30)
print("Stok beras setelah penjualan:",
      komoditas_list[0].stok_kg, "kg")

# LO 4: Uji validasi input
print("\n=== UJI VALIDASI ===")
try:
    komoditas_salah = KomoditasPertanian("Beras", -5000, 100)
except ValueError as error:
    print("Kesalahan:", error)

try:
    komoditas_salah = KomoditasPertanian("Jagung", 8000, -10)
except ValueError as error:
    print("Kesalahan:", error)

# LO 4: Getter, modifier, dan dunder method
print("\n=== GETTER, MODIFIER, DUNDER ===")
print("Nilai stok beras:",
      komoditas_list[0].hitung_nilai_stok())

komoditas_list[0].tambah_stok(10)
print("Stok setelah dimodifikasi:",
      komoditas_list[0].stok_kg, "kg")

print("Representasi objek:", str(komoditas_list[0]))

# LO 5: Atribut kelas
print("\n=== ATRIBUT KELAS ===")
print("Jumlah objek:",
      KomoditasPertanian.jumlah_objek)

komoditas_baru = KomoditasPertanian("Cabai", 25000, 40)
print("Jumlah setelah objek baru dibuat:",
      KomoditasPertanian.jumlah_objek)

print("Atribut kelas melalui kelas:",
      KomoditasPertanian.diskon_persen)
print("Atribut kelas melalui objek:",
      komoditas_list[0].diskon_persen)

# Mengubah atribut kelas dan melihat efeknya
KomoditasPertanian.diskon_persen = 10

print("Harga beras setelah diskon:",
      komoditas_list[0].hitung_harga_diskon())
print("Harga jagung setelah diskon:",
      komoditas_list[1].hitung_harga_diskon())

# LO 6: Kelas perangkat fisik agroindustri
class MesinSortasi:
    """Merepresentasikan mesin sortasi hasil pertanian."""

    def __init__(self, kapasitas_maks_kg: float,
                 suhu_maks_c: float):
        """Membuat mesin dengan batas kapasitas dan suhu."""
        if kapasitas_maks_kg <= 0 or suhu_maks_c <= 0:
            raise ValueError("Batas mesin harus positif.")

        self.kapasitas_maks_kg = kapasitas_maks_kg
        self.suhu_maks_c = suhu_maks_c

    def jalankan_sortasi(self, berat_kg: float,
                         suhu_c: float) -> str:
        """Menolak operasi jika batas kapasitas atau suhu dilanggar."""
        if berat_kg <= 0:
            raise ValueError("Berat harus lebih dari nol.")
        if berat_kg > self.kapasitas_maks_kg:
            raise ValueError("Berat melebihi kapasitas mesin.")
        if suhu_c < 0 or suhu_c > self.suhu_maks_c:
            raise ValueError("Suhu berada di luar batas mesin.")

        return (
            f"Sortasi berhasil: {berat_kg} kg pada "
            f"suhu {suhu_c} derajat C."
        )


print("\n=== UJI MESIN SORTASI ===")
mesin = MesinSortasi(200, 40)

try:
    print(mesin.jalankan_sortasi(150, 30))
except ValueError as error:
    print("Kesalahan:", error)

try:
    print(mesin.jalankan_sortasi(250, 30))
except ValueError as error:
    print("Kesalahan:", error)

try:
    print(mesin.jalankan_sortasi(100, 50))
except ValueError as error:
    print("Kesalahan:", error)
```
